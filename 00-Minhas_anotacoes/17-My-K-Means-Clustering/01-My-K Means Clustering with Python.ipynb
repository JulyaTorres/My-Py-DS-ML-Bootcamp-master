```python
# ==========================================
# 1. IMPORTAÇÃO DE BIBLIOTECAS
# ==========================================
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
# from sklearn.preprocessing import StandardScaler # (Descomente para dados reais)
```

```python
# Configuração estética dos gráficos
sns.set_theme(style="darkgrid")
```

```python
# ==========================================
# 2. CRIAÇÃO E PREPARAÇÃO DOS DADOS
# ==========================================
# Desempacotamento limpo: X recebe as coordenadas, y recebe os grupos reais
X, y = make_blobs(n_samples=200, 
                  n_features=2, 
                  centers=4, 
                  cluster_std=1.8, 
                  random_state=101)
```

```python
# DICA: Se fossem dados reais, você faria a padronização aqui:
# scaler = StandardScaler()
# X_scaled = scaler.fit_transform(X)
```

```python
# ==========================================
# 3. TREINAMENTO DO MODELO K-MEANS
# ==========================================
# n_init='auto' é a recomendação atual para suprimir alertas do Scikit-Learn
kmeans = KMeans(n_clusters=4, n_init='auto', random_state=42)
kmeans.fit(X) # Se usar StandardScaler, mude para fit(X_scaled)
```

```python
# Extraindo os resultados
centroides = kmeans.cluster_centers_
rotulos_previstos = kmeans.labels_
```

```python
# ==========================================
# 4. VISUALIZAÇÃO E INTERPRETAÇÃO DOS RESULTADOS
# ==========================================
# Corrigido o uso de ax1 e ax2
fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True, figsize=(12, 6))

# ---- Gráfico 1: Agrupamento feito pelo Algoritmo ----
ax1.set_title('Agrupamento Previsto (K-Means)', fontsize=14)
ax1.scatter(X[:, 0], X[:, 1], c=rotulos_previstos, cmap='rainbow', edgecolor='k', alpha=0.7)
# Plotando os centroides encontrados para maior clareza visual
ax1.scatter(centroides[:, 0], centroides[:, 1], color='black', marker='X', s=150, linewidths=3, label='Centroides')
ax1.legend()

# ---- Gráfico 2: Divisão Original (Gabarito) ----
ax2.set_title('Grupos Originais (Reais)', fontsize=14)
ax2.scatter(X[:, 0], X[:, 1], c=y, cmap='rainbow', edgecolor='k', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 5. DESCOBRINDO O "K" IDEAL (ELBOW METHOD)
# ==========================================
# Lista vazia para armazenar a inércia de cada iteração
valores_inercia = []

# Definindo o intervalo de clusters que queremos testar (ex: de 1 a 10)
intervalo_k = range(1, 11)

# Loop: Treina um modelo novo para cada valor de K e salva a inércia
for k in intervalo_k:
    # Instancia o modelo com o K da vez
    kmeans_temp = KMeans(n_clusters=k, n_init='auto', random_state=42)
    kmeans_temp.fit(X) # Usa X_scaled se estiver lidando com dados reais

    # A propriedade .inertia_ retorna o WCSS do modelo treinado
    valores_inercia.append(kmeans_temp.inertia_)

# ==========================================
# Visualizando o Gráfico do Cotovelo
# ==========================================
plt.figure(figsize=(9, 5))
plt.plot(intervalo_k, valores_inercia, marker='o', linestyle='--', color='#2ca02c', linewidth=2, markersize=8)

plt.title('Método do Cotovelo (Elbow Method)', fontsize=15, pad=15)
plt.xlabel('Número de Clusters (K)', fontsize=12)
plt.ylabel('Inércia (WCSS)', fontsize=12)

# Força o eixo X a mostrar apenas números inteiros (1, 2, 3...)
plt.xticks(intervalo_k)

# Opcional: Adicionando uma marcação visual (Sabemos que o K ideal dos nossos dados é 4)
plt.axvline(x=4, color='red', linestyle=':', label='Possível Cotovelo (K=4)')
plt.legend()

plt.tight_layout()
plt.show()
```
