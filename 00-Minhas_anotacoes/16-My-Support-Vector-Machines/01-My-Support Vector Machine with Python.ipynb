# Correção do Código

## Importando Bibliotecas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Carregando os Dados

```python
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer(as_frame=True)
df = cancer.frame
df.head()
```

## Entendo o alvo (`target`)

```python
print("Classes:", dict(enumerate(cancer.target_names)))
print()
print(df['target'].value_counts())
print()
print(df['target'].value_counts(normalize=True).round(3))
```

## Divisão em treino e teste

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns='target')
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=101,
    stratify=y
)

print("Treino:", X_train.shape, "| Teste:", X_test.shape)
print()
print("Proporção no treino:\n", y_train.value_counts(normalize=True).round(3))
print()
print("Proporção no teste:\n", y_test.value_counts(normalize=True).round(3))
```

## Treinando o SVC "cru" (sem normalizar)

```python
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report

model = SVC()
model.fit(X_train, y_train)

predictions = model.predict(X_test)

print(confusion_matrix(y_test, predictions))
print()
print(classification_report(y_test, predictions, target_names=cancer.target_names))
```

## Reproduzindo o problema do Curso (`gamma='auto'`)

```python
model_auto = SVC(gamma='auto')
model_auto.fit(X_train, y_train)
pred_auto = model_auto.predict(X_test)

print(confusion_matrix(y_test, pred_auto))
print()
print(classification_report(y_test, pred_auto, target_names=cancer.target_names, zero_division=0))
```

## Normalizando os dados (StandardScaler)

```python
from sklearn.preprocessing import StandardScaler

# Antes: escalas muito diferentes entre as features
print("Desvio padrão ANTES (treino):")
print(X_train[['mean area', 'mean perimeter', 'mean smoothness']].std().round(3))

scaler = StandardScaler().set_output(transform="pandas")
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Depois: todas com média ~0 e desvio padrão ~1
print("\nDesvio padrão DEPOIS (treino):")
print(X_train_scaled[['mean area', 'mean perimeter', 'mean smoothness']].std().round(3))
print("\nMédia DEPOIS (treino):")
print(X_train_scaled[['mean area', 'mean perimeter', 'mean smoothness']].mean().round(3))
```

## Treinando o SVC com dados normalizados

```python
model_scaled = SVC()
model_scaled.fit(X_train_scaled, y_train)

pred_scaled = model_scaled.predict(X_test_scaled)

print(confusion_matrix(y_test, pred_scaled))
print()
print(classification_report(y_test, pred_scaled, target_names=cancer.target_names))
```

## GridSearchCV com Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV, StratifiedKFold

pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC())
])

param_grid = {
    'svc__C': [0.1, 1, 10, 100],
    'svc__gamma': [0.001, 0.01, 0.1, 1],
    'svc__kernel': ['rbf']
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=101)

grid = GridSearchCV(pipe, param_grid, cv=cv, refit=True, n_jobs=-1, verbose=1)
grid.fit(X_train, y_train)

print("Melhores parâmetros:", grid.best_params_)
print("Melhor acurácia na validação cruzada:", round(grid.best_score_, 4))
```

## Avaliação final no conjunto de teste

```python
grid_predictions = grid.predict(X_test)

print(confusion_matrix(y_test, grid_predictions))
print()
print(classification_report(y_test, grid_predictions, target_names=cancer.target_names))
```

## Visualizando a grade (heatmap de C × gamma)

```python
results = pd.DataFrame(grid.cv_results_)
results['param_svc__C'] = results['param_svc__C'].astype(float)
results['param_svc__gamma'] = results['param_svc__gamma'].astype(float)

pivot = results.pivot_table(
    index='param_svc__C',
    columns='param_svc__gamma',
    values='mean_test_score'
)

plt.figure(figsize=(7, 5))
sns.heatmap(pivot, annot=True, fmt='.3f', cmap='viridis')
plt.title('Acurácia média na validação cruzada (5 folds)')
plt.xlabel('gamma')
plt.ylabel('C')
plt.show()
```

## Matriz de confusão visual

```python
from sklearn.metrics import ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

ConfusionMatrixDisplay.from_estimator(
    grid, X_test, y_test,
    display_labels=cancer.target_names,
    cmap='Blues', ax=axes[0]
)
axes[0].set_title('Contagem')

ConfusionMatrixDisplay.from_estimator(
    grid, X_test, y_test,
    display_labels=cancer.target_names,
    normalize='true', values_format='.2f',
    cmap='Blues', ax=axes[1]
)
axes[1].set_title('Normalizada por linha (= recall)')

for ax in axes:
    ax.grid(False)

plt.tight_layout()
plt.show()
```

## Fronteira de decisão em 2D

```python
from sklearn.inspection import DecisionBoundaryDisplay

feats = ['worst radius', 'worst concave points']

scaler_2d = StandardScaler()
X2_train = scaler_2d.fit_transform(X_train[feats])

svc_2d = SVC(C=10, gamma='scale')
svc_2d.fit(X2_train, y_train)

fig, ax = plt.subplots(figsize=(7, 5))

DecisionBoundaryDisplay.from_estimator(
    svc_2d, X2_train,
    response_method='predict',
    cmap='coolwarm_r', alpha=0.3, ax=ax,
    xlabel=f'{feats[0]} (padronizado)',
    ylabel=f'{feats[1]} (padronizado)'
)

for classe, cor in [(0, 'tab:red'), (1, 'tab:blue')]:
    mask = (y_train == classe).to_numpy()
    ax.scatter(X2_train[mask, 0], X2_train[mask, 1], c=cor, s=25,
               edgecolor='k', label=cancer.target_names[classe])

ax.scatter(svc_2d.support_vectors_[:, 0], svc_2d.support_vectors_[:, 1],
           s=130, facecolors='none', edgecolors='k', linewidths=1.2,
           label='vetores de suporte')

ax.legend()
ax.set_title('SVC (kernel RBF) com 2 features')
plt.show()

print("Vetores de suporte por classe:", svc_2d.n_support_)
```

## O efeito de gamma e C na fronteira

```python
def plota_fronteira(ax, C, gamma):
    m = SVC(C=C, gamma=gamma).fit(X2_train, y_train)
    DecisionBoundaryDisplay.from_estimator(
        m, X2_train, response_method='predict',
        cmap='coolwarm_r', alpha=0.3, ax=ax, xlabel='', ylabel=''
    )
    ax.scatter(X2_train[:, 0], X2_train[:, 1], c=y_train,
               cmap='coolwarm_r', s=15, edgecolor='k', linewidth=0.3)
    acc = m.score(X2_train, y_train)
    ax.set_title(f'C={C}, gamma={gamma}\nSV={m.n_support_.sum()} | acurácia treino={acc:.3f}',
                 fontsize=10)
    ax.grid(False)

fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Linha de cima: C fixo, gamma variando
for ax, g in zip(axes[0], [0.01, 0.5, 20]):
    plota_fronteira(ax, C=10, gamma=g)

# Linha de baixo: gamma fixo, C variando
for ax, c in zip(axes[1], [0.01, 1, 1000]):
    plota_fronteira(ax, C=c, gamma=0.5)

plt.tight_layout()
plt.show()
```

## Comparando kernels (com as 30 features)

```python
from sklearn.model_selection import cross_validate
from sklearn.metrics import make_scorer, recall_score

scoring = {
    'acuracia': 'accuracy',
    'recall_malignant': make_scorer(recall_score, pos_label=0)
}

modelos = {
    'linear (C=1)': SVC(kernel='linear', C=1),
    'poly grau 3 (C=1)': SVC(kernel='poly', degree=3, C=1),
    'sigmoid (C=1)': SVC(kernel='sigmoid', C=1),
    'rbf padrão (C=1)': SVC(kernel='rbf', C=1),
    'rbf ajustado (C=10, gamma=0.01)': SVC(kernel='rbf', C=10, gamma=0.01),
}

linhas = []
for nome, svc in modelos.items():
    pipe_k = Pipeline([('scaler', StandardScaler()), ('svc', svc)])
    res = cross_validate(pipe_k, X_train, y_train, cv=cv, scoring=scoring)
    linhas.append({
        'kernel': nome,
        'acurácia': res['test_acuracia'].mean(),
        'recall malignant': res['test_recall_malignant'].mean()
    })

pd.DataFrame(linhas).set_index('kernel').round(3).sort_values('acurácia', ascending=False)
```

## Priorizando o recall de malignant com class_weight

```python
from sklearn.metrics import precision_score

scoring = {
    'acuracia': 'accuracy',
    'recall malignant': make_scorer(recall_score, pos_label=0),
    'precision malignant': make_scorer(precision_score, pos_label=0),
}

modelos_cw = {
    'rbf ajustado': SVC(C=10, gamma=0.01),
    'rbf ajustado + class_weight': SVC(C=10, gamma=0.01, class_weight='balanced'),
    'linear': SVC(kernel='linear', C=1),
    'linear + class_weight': SVC(kernel='linear', C=1, class_weight='balanced'),
}

linhas = []
for nome, svc in modelos_cw.items():
    pipe_cw = Pipeline([('scaler', StandardScaler()), ('svc', svc)])
    res = cross_validate(pipe_cw, X_train, y_train, cv=cv, scoring=scoring)
    linhas.append({
        'modelo': nome,
        'acurácia': res['test_acuracia'].mean(),
        'recall malignant': res['test_recall malignant'].mean(),
        'precision malignant': res['test_precision malignant'].mean(),
    })

pd.DataFrame(linhas).set_index('modelo').round(3)
```
