```python
def text_process(mess):
    """
    Takes in a string of text, then performs the following:
    1. Remove all punctuation
    2. Remove all stopwords
    3. Returns a list of the cleaned text
    """
    # Check characters to see if they are in punctuation
    nopunc = [char for char in mess if char not in string.punctuation]

    # Join the characters again to form the string.
    nopunc = ''.join(nopunc)
    
    # Now just remove any stopwords
    return [word for word in nopunc.split() if word.lower() not in stopwords.words('english')]
```

```python
# ==========================================
# 1. IMPORTAÇÃO DE BIBLIOTECAS
# ==========================================
import pandas as pd
import string
import nltk
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
# Baixa a lista de stopwords (Execute apenas uma vez no seu ambiente)
# nltk.download('stopwords', quiet=True)
```

```python
# ==========================================
# 2. FUNÇÃO DE LIMPEZA DE TEXTO (PRÉ-PROCESSAMENTO)
# ==========================================
def processar_texto(mensagem):
    """
    1. Remove pontuações
    2. Remove stopwords (palavras vazias em inglês, ou 'portuguese' se o texto for do Brasil)
    3. Retorna uma lista de palavras limpas
    """
    # Compreensão de lista para ignorar caracteres que são pontuação
    sem_pontuacao = [char for char in mensagem if char not in string.punctuation]
    sem_pontuacao = ''.join(sem_pontuacao)

    # Filtra as palavras vazias e retorna a lista final
    return [palavra for palavra in sem_pontuacao.split()
            if palavra.lower() not in stopwords.words('english')]
```

```python
# ==========================================
# 3. CARREGAMENTO E DIVISÃO DOS DADOS
# ==========================================
# Lendo os dados (supondo que o arquivo seja SMSSpamCollection)
mensagens = pd.read_csv('smsspamcollection/SMSSpamCollection', sep='\t', names=['label', 'message'])

# Divisão em Treino e Teste (Corrigindo o erro de digitação do material)
msg_treino, msg_teste, rotulo_treino, rotulo_teste = train_test_split(
    mensagens['message'],
    mensagens['label'],
    test_size=0.2,
    random_state=101
)
```

```python
# ==========================================
# 4. CRIANDO A ESTEIRA (PIPELINE) AUTOMATIZADA
# ==========================================
# O Pipeline aplica as etapas na ordem correta, automatizando o fluxo de dados.
# Isso substitui todas aquelas linhas soltas do material original.
esteira_pln = Pipeline([
    ('bag_of_words', CountVectorizer(analyzer=processar_texto)), # Conta as palavras
    ('tfidf', TfidfTransformer()),                               # Calcula os pesos (Frequência/Raridade)
    ('classificador', MultinomialNB())                           # Treina o modelo Naive Bayes
])
```

```python
# ==========================================
# 5. TREINAMENTO E AVALIAÇÃO
# ==========================================
# Apenas chamar o .fit() na esteira já roda todas as 3 etapas matemáticas
esteira_pln.fit(msg_treino, rotulo_treino)

# Avaliando com os dados de teste que o modelo nunca viu
previsoes = esteira_pln.predict(msg_teste)

print("Relatório de Classificação (Naive Bayes):")
print(classification_report(rotulo_teste, previsoes))
```
