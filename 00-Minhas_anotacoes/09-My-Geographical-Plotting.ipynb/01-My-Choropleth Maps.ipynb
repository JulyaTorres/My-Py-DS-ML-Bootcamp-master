# Bibliotecas

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.offline import init_notebook_mode, iplot

# Permite exibir gráficos no Jupyter Notebook sem precisar de conexão com a nuvem
init_notebook_mode(connected=True)
```

```python
data = dict(
    type='choropleth',
    locations=['AZ', 'CA', 'NY'],
    locationmode='USA-states',
    colorscale='Portland',
    text=['Texto AZ', 'Texto CA', 'Texto NY'],
    z=[1.0, 2.0, 3.0],
    colorbar={'title': 'Legenda de Cores'}
)

layout = dict(geo={'scope': 'usa'})  # Define que o mapa será dos EUA

fig = go.Figure(data=[data], layout=layout)
iplot(fig)
```

```python
df = pd.read_csv('2011_US_AGRI_Exports')
df.head()
```

```python
data = dict(
    type='choropleth',
    colorscale='ylorrd',  # Amarelo → Vermelho
    locations=df['code'],  # Siglas dos estados
    z=df['total exports'],  # Valores que definem a cor
    locationmode='USA-states',
    text=df['text'],  # Tooltip
    marker=dict(
        line=dict(color='rgb(255,255,255)', width=2)  # Linhas divisórias brancas
    ),
    colorbar={'title': "Milhões USD"}
)

layout = dict(
    title='2011 US Agriculture Exports by State',
    geo=dict(
        scope='usa',
        showlakes=True,
        lakecolor='rgb(85,173,240)'  # Cor dos lagos
    )
)

fig = go.Figure(data=[data], layout=layout)
iplot(fig)
```

```python
df = pd.read_csv('2014_World_GDP')
df.head()
```

```python
data = dict(
    type='choropleth',
    locations=df['CODE'],  # Códigos ISO de 3 letras
    z=df['GDP (BILLIONS)'],
    text=df['COUNTRY'],
    colorbar={'title': 'PIB em Bilhões US$'}
)

layout = dict(
    title='2014 Global GDP',
    geo=dict(
        showframe=False,  # Remove a moldura do mapa
        projection={'type': 'mollweide'}  # Tipo de projeção
    )
)

fig = go.Figure(data=[data], layout=layout)
iplot(fig)
```
