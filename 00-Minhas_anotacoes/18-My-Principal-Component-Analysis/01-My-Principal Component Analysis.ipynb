```python
# ==========================================
# 1. IMPORTAÇÃO DE BIBLIOTECAS
# ==========================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
sns.set_theme(style="darkgrid")
```

```python
# ==========================================
# 2. CARREGAMENTO E EXPLORAÇÃO (Forma Moderna)
# ==========================================
# O parâmetro as_frame=True retorna direto um DataFrame do Pandas, evitando dicionários
dados_cancer = load_breast_cancer(as_frame=True)

# X recebe todas as features (as 30 colunas numéricas)
X = dados_cancer.data
# y recebe os rótulos (benigno ou maligno)
y = dados_cancer.target
nomes_alvo = dados_cancer.target_names # ['malignant', 'benign']
```

```python
# ==========================================
# 3. PRÉ-PROCESSAMENTO: PADRONIZAÇÃO (OBRIGATÓRIO PARA PCA)
# ==========================================
# Sem isso, variáveis com grande magnitude esmagam o cálculo do PCA
scaler = StandardScaler()
X_escalonado = scaler.fit_transform(X)
```

```python
# ==========================================
# 4. APLICAÇÃO DO PCA
# ==========================================
# Definimos que queremos reduzir as 30 colunas originais para apenas 2 componentes
pca = PCA(n_components=2)
# O modelo aprende os padrões e já aplica a transformação
X_pca = pca.fit_transform(X_escalonado)

# Verificando quanta informação conseguimos manter
variancia_explicada = pca.explained_variance_ratio_
print(f"Variância mantida pelo PC1: {variancia_explicada[0]*100:.2f}%")
print(f"Variância mantida pelo PC2: {variancia_explicada[1]*100:.2f}%")
print(f"Total de informação preservada: {sum(variancia_explicada)*100:.2f}%")
```

```python
# ==========================================
# 5. VISUALIZAÇÃO: SCATTER PLOT DOS COMPONENTES
# ==========================================
# Primeiro, mapeamos os números (0 e 1) para os nomes reais do dataset ('malignant' e 'benign')
y_nomes = [nomes_alvo[i] for i in y]

plt.figure(figsize=(10, 6))

# Agora, passamos a nova lista com os nomes de texto diretamente no 'hue'
grafico = sns.scatterplot(x=X_pca[:, 0], 
                          y=X_pca[:, 1], 
                          hue=y_nomes, # Utilizar as strings aqui resolve o problema da legenda
                          palette='Set1', 
                          alpha=0.8)

plt.title('Separação dos Dados usando apenas 2 Componentes Principais', fontsize=14)
plt.xlabel('Primeiro Componente Principal (PC1)')
plt.ylabel('Segundo Componente Principal (PC2)')

# Como passámos os nomes diretamente no 'hue', o Seaborn já associou as cores corretamente.
# Precisamos apenas de definir o título da legenda.
plt.legend(title='Diagnóstico')

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 6. BÔNUS: O QUE HÁ DENTRO DOS COMPONENTES? (HEATMAP)
# ==========================================
# pca.components_ mostra o peso estatístico que cada variável original teve na criação do PC1 e PC2
df_componentes = pd.DataFrame(pca.components_, columns=X.columns, index=['PC1', 'PC2'])

plt.figure(figsize=(14, 4))
sns.heatmap(df_componentes, cmap='coolwarm', annot=False)
plt.title('Correlação: Variáveis Originais vs Componentes Principais', fontsize=14)
plt.xlabel('Features Originais do Dataset')
plt.ylabel('Componentes')
plt.tight_layout()
plt.show()
```
