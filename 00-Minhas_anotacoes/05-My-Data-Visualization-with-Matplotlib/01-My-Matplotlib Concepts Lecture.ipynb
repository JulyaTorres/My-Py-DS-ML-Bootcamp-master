# Importando Bibliotecas

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
%matplotlib inline
```

# Exemplo

```python
x = np.linspace(0,5,11)
y = x ** 2
```

```python
x
```

```python
y
```

# comandos básicos

```python
# FUNCTIONAL
plt.plot(x,y)
```

```python
# Alternado cor da linha
plt.plot(x,y, 'red')
```

```python
# Mais detalhes
plt.plot(x,y, 'red') # gráfico básico com linha vermelha
plt.xlabel('X Label') # nome da variavel x
plt.ylabel('Y Label') # nome da variavel y
plt.title('Title')    # título do gráfico
```

## Multiplots na mesma tela

```python
# plt.subplot(nrows, ncols, plot_number)

#subplot 1
plt.subplot(1,2,1)
plt.plot(x,y,'red')

# subplot 2
plt.subplot(1,2,2)
plt.plot(y,x,'purple')
```

# Método Orientado a Objetos

```python
# OO

# Cria figura (tela vazia)
fig = plt.figure()

# Adiciona conjunto de eixos à figura
axes = fig.add_axes([0.1,0.1,0.8,0.8]) # [esquerda, baixo, largura, altura] em porcentagem (range 0 to 1)

# Trace o gráfico nesse conjunto de eixos
axes.plot(x,y)
axes.set_xlabel('X Label')
axes.set_ylabel('Y Label')
axes.set_title('Set Title')
```

```python
# Criando figura em branco
fig = plt.figure()

# Adiciona conjunto de eixos à figura
axes1 = fig.add_axes([0.1,0.1,0.8,0.8]) # main axes
axes2 = fig.add_axes([0.2,0.5,0.4,0.3]) # inset axes

# Gráfico maior 1 
axes1.plot(x,y, 'b')
axes1.set_xlabel('X_label_axes1')
axes1.set_ylabel('Y_label_axes1')
axes1.set_title('LARGER PLOT')

# Gráfico pequeno dentro do maior
axes2.plot(y,x,'r')
axes2.set_xlabel('X_label_axes2')
axes2.set_ylabel('Y_label_axes2')
axes2.set_title('SALLER PLOT')
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2)

#axes.plot(x,y)

for current_ax in axes:
    current_ax.plot(x,y)
```

# Subplots()

gerenciador de eixos mais automáticos

```python
fig,axes = plt.subplots(nrows=1,ncols=2)

axes[0].plot(x,y, 'r')
axes[0].set_title('First Plot')


axes[1].plot(y,x,'b')
axes[1].set_title('Second Title')
```

```python
fig = plt.figure(figsize = (3,2))

ax = fig.add_axes([0,0,1,1])
ax.plot(x,y)
```
