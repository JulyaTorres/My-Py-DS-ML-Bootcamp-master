# Atalhos Jupyter Notebook

## Criando células de código

- `shift + Enter` -> Executa a célula e cria uma nova abaixo
- `Ctrl + Enter` -> Executa a célula sem criar outra.

## Criando células de texto (Markdown)

- **Clique na célula** e troque e tipo no menu superior (`Code -> Markdown`)
- **Atalho**: Use `Esc + M` para mudar uma célula para Markdown
- Escreva o texto normalmente, usando **Markdown** para formatação:
    - "#" Título Grande
    - "##" Subtítulo
    - "**palavra**", "*itálico*", "`código embutido`"
- **Executar**: Pressione `Shift + Enter` para renderizar o texto

## Personalizar ou adcionar novos atalhos

- Clique no **lápis** (ícone de edição) ao lado do atalho que deseja mudar
- Pressione a nova combinação de teclas que deseja usar
- Pressione **Enter** para salvar
