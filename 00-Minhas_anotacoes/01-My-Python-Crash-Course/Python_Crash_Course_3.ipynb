# for Loops
O loop for permite **iterar sobre uma sequência** de elementos (como listas, strings ou tuplas).

```python
# Criando uma lista
seq = [1,2,3,4,5]
```

```python
# Para cada item em 'seq' impima item
for item in seq: # O for pecorre a lista seq, um item por vez
    print(item) # imprime cada item
```

```python
# Para cada item em seq (5) imprima 'hello'
for num in seq:
    print('hello')
```

 Lembrando: pode usar qualquer palavra ou letra depois do for

# While Loops
Excuta um bloco enquanto a confição for verdadeira:

```python
i = 1 # inicia i com o valor 1

while i < 5: # enquanto i for menor qe 5, o bloco será executado
    print('i is: {}'.format(i)) # usa a formatação com .format() para inserir o valor i na string
    i = i+1 # após imprimir, i é incrementado em 1
```

```python
for x in seq: # para cada x/elemnto em seq imprima x/elemento
    print(x)
```

# range()
Gera uma sequência de números de 0 até n-1:

```python
# Cria um intervalo que representa: 0, 1, 2, 3, 4
range(0,5)
```

```python
for x in range(0, 5): # # itera sobre os valores gerados por range(5)
    print(x) # # imprime os números de 0 a 4
```

```python
list(range(0,5)) # Converte o intervalo em uma lista: [0, 1, 2, 3, 4]
```

```python
list(range(10))
```

# list comprehension
Forma compacta de criar listas a partir de outra:

```python
x = [1,2,3,4]
```

```python
# Forma Tradicional

out = [] # inicializa uma lista vazia chamada out

for num in x: # o loop passará por cada item na lista x
    out.append(num**2) # calcula item**2 para cada item em x e adiciona em out
```

```python
print(out)
```

```python
# Comprehension
[num**2 for num in x]
```

```python
# Criando a lista em uma variável com comprehension
out = [num ** 2 for num in x ]
```

```python
out
```

# Funções
Definem blocos reutilizáveis que podem aceitar entradas (parâmetros) e retornar resultados.

```python
def my_func(param1): # 'def' defini uma função chamada 'my_func', 'param1' é um parâmetro com valor padrão 'default'
    """
    Docstring goes here. 
    # É um comentário explicativo que fica visível com help(my_func)
    """
    print(param1) # imprime o valor fornecido
```

## Chamando a função

```python
# apenas mostra que a função existe
my_func
```

```python
# chama com valor padrão -> imprime 'default'
my_func()
```

```python
# imprime 'hello'
my_func('hello')
```

```python
# forma nomeada (igual resultado)
my_func(param1='hello')
```

## Mais exemplos de funções

```python
my_func('Jose')
```

```python
my_func(name = 'Jose')
```
