# Importando Bibliotecas

```python
import numpy as np
import pandas as pd
```

# Criando DataFrame

```python
# Create dataframe
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

# Agrupamento

```python
# Agrupando os dados com base na coluna 'Company'
df.groupby('Company')
```

```python
# Salvando agrupamento como variável
byComp = df.groupby('Company')
```

```python
byComp
```

```python
# chame métodos agregados do objeto -> média das companhias somente com os número, no caso 'Sales'
byComp.mean(numeric_only=True)
```

```python
# Soma das vendas de cada companhia
byComp.sum()
```

```python
# Desvio padrão
byComp.std(numeric_only=True)
```

```python
byComp.sum().loc['FB']
```

```python
# Soma da variavel numerica, no caso 'Sales', da companhia 'FB'
df.groupby('Company').sum(numeric_only=True).loc['FB']
```

```python
# Contagem
df.groupby('Company').count()
```

```python
# Máximo
df.groupby('Company').max()
```

```python
df
```

```python
# Mínimo
df.groupby('Company').min()
```

```python
# Estatísticas completas
df.groupby('Company').describe()
```

```python
# Estatísticas completas transpostas
df.groupby('Company').describe().transpose()
```

```python
# Estatísticas completas transpostas de 'FB'
df.groupby('Company').describe().transpose()['FB']
```

```python
# Estatísticas completas de 'FB'
df.groupby('Company').describe().loc['FB']
```
