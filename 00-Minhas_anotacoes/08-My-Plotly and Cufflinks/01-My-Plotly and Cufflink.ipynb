```python
import pandas as pd
import numpy as np
import plotly.express as px

import plotly
print("Versão do Plotly:", plotly.__version__)
```

```python
# DataFrame com valores aleatórios
df = pd.DataFrame(np.random.randn(100, 4), columns=list("ABCD"))
df.head()

# DataFrame simples para barras
df2 = pd.DataFrame({
    'Category': ['A', 'B', 'C'],
    'Values': [32, 43, 50]
})
df2.head()
```

```python
df.head()
```

```python
fig = px.scatter(df, x='A', y='B', title='Scatter Plot', color_discrete_sequence=['pink'])
fig.show()
```

```python
# Barras simples
fig = px.bar(df2, x='Category', y='Values', title='Bar Plot - Plotly Express')
fig.show()
```

```python

# Crie um DataFrame de exemplo com dados aleatórios
dados = {
    'a': ['Categoria 1', 'Categoria 2', 'Categoria 3', 'Categoria 4'],
    'b': [10, 25, 13, 30]
}
df = pd.DataFrame(dados)

df.head()
```

```python
# Agora, o seu código para o gráfico funcionará
fig = px.bar(df, x='a', y='b', title='Barras de A x B (Exemplo Aleatório)')
fig.show()
```

```python
# Crie o gráfico de barras, usando a coluna 'a' para os nomes das barras (x)
# e também para a cor de cada barra (color)
fig = px.bar(df, x='a', y='b', title='Barras de A x B com cores diferentes', color='a')

fig.show()
```

```python
df = pd.DataFrame(np.random.randn(100, 4), columns=['A','B','C','D'])
df_long = df.melt(var_name='coluna', value_name='valor')

df_long.head()
```

```python
# 2. Crie o boxplot usando o novo DataFrame
fig = px.box(
    df_long,
    x='coluna',       # O eixo X agora são as categorias (A, B, C, D)
    y='valor',        # O eixo Y são os valores numéricos
    color='coluna',   # Use a nova coluna 'coluna' para colorir as caixas
    title='Boxplot das colunas (com cores diferentes)'
)
fig.show()
```

```python
import plotly.graph_objects as go
import pandas as pd

df3 = pd.DataFrame({'x':[1,2,3,4,5],'y':[10,20,30,20,10],'z':[5,4,3,2,1]})

# Para um gráfico de superfície, normalmente z é uma matriz (grid). 
# Com seus dados, podemos fazer um gráfico de dispersão 3D:
fig = go.Figure(data=[go.Scatter3d(
    x=df3['x'],
    y=df3['y'],
    z=df3['z'],
    mode='markers',
    marker=dict(size=8, color=df3['z'], colorscale='magma')
)])
fig.update_layout(title='Dispersão 3D')
fig.show()
```

```python
# Exemplo: criar grid
x = np.linspace(0, 10, 30)
y = np.linspace(0, 10, 30)
x_grid, y_grid = np.meshgrid(x, y)

# Definir Z como alguma função de X e Y
z_grid = np.sin(x_grid) + np.cos(y_grid)

# Criar gráfico
fig = go.Figure(data=[go.Surface(z=z_grid, x=x_grid, y=y_grid, colorscale='magma')])
fig.update_layout(title='Superfície 3D')
fig.show()
```

```python
# DataFrame com valores aleatórios
df = pd.DataFrame(np.random.randn(100, 4), columns=list("ABCD"))
df.head()
```

```python
fig = px.histogram(df, x='A', nbins=25, title='Histograma da coluna A', color_discrete_sequence=['pink'])
fig.show()
```

```python
df['C'] = df['C'] - df['C'].min() + 1  # desloca tudo para ficar positivo

fig = px.scatter(df, x='A', y='B', size='C', color='D', title='Gráfico de Bolhas')
fig.show()
```

```python
fig = px.scatter_matrix(df, dimensions=['A', 'B', 'C', 'D'], title='Scatter Matrix', color_discrete_sequence=['red'])
fig.show()
```
