# Importando Biliotecas

```python
import seaborn as sns
import numpy as np
%matplotlib inline 
```

# Importando Datasets

```python
tips = sns.load_dataset('tips')
tips.head()
```

# barplot e countplot

```python
sns.barplot(x= 'sex',y= 'total_bill',data=tips, palette='Set1')
```

```python
sns.barplot(x= 'sex',y= 'total_bill',data=tips, estimator=np.std, palette ='Set1')
```

## Countplot

```python
sns.countplot(x='sex',data=tips, palette ='Set1')
```

## boxplot e violinplot

```python
sns.boxplot(x='day',y='total_bill',data=tips, palette ='rainbow')
```

```python
sns.boxplot(x='day',y='total_bill',data=tips, hue='smoker', palette ='rainbow')
```

### violinplot

```python
sns.violinplot(x='day',y='total_bill',data=tips, palette ='rainbow')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips, hue='sex', palette ='Set1')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips, hue='sex', split=True, palette ='Set1')
```

## stripplot e swarmplot

```python
sns.stripplot(x='day',y='total_bill',data=tips, palette ='rainbow')
```

```python
sns.stripplot(x='day',y='total_bill',data=tips, jitter=True, palette ='rainbow')
```

```python
sns.stripplot(x='day',y='total_bill',data=tips, jitter=True, hue='sex')
```

```python
sns.stripplot(x='day',y='total_bill',data=tips, jitter=True, hue='sex')
```

### swarmplot

```python
sns.swarmplot(x='day', y='total_bill',data=tips, palette ='rainbow')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips, color='purple')
sns.swarmplot(x='day', y='total_bill',data=tips, color='pink')
```

## catplot

```python
sns.catplot(x='day', y='total_bill', data=tips, kind='bar', palette ='rainbow')
```

```python
sns.catplot(x='day', y='total_bill', data=tips, kind='violin', palette ='rainbow')
```
